```python
import pandas as pd
from pathlib import Path

# Look for the uploaded file 'train (1).csv' or 'train.csv'
input_path = Path('train (1).csv')
if not input_path.exists():
    input_path = Path('train.csv')

if not input_path.exists():
    raise FileNotFoundError("Place your original dataset in this folder and name it train.csv or train (1).csv")

# Reload the original dataset
df = pd.read_csv(input_path)

# Remove duplicate rows
df = df.drop_duplicates()

# Fill missing Age values with median
df["Age"] = df["Age"].fillna(df["Age"].median())

# Fill missing Embarked values with mode
df["Embarked"] = df["Embarked"].fillna(df["Embarked"].mode()[0])

# Create an indicator for whether cabin information is available
df["CabinKnown"] = df["Cabin"].notna().astype(int)

# Drop Cabin because it has many missing values
df = df.drop(columns=["Cabin"])

# Remove unnecessary spaces from text columns
for col in df.select_dtypes(include="object").columns:
    df[col] = df[col].str.strip()

print("Dataset cleaned successfully from:", input_path)
```

```python
print("Missing values after cleaning:")
print(df.isnull().sum())
```

```python
print("Duplicate rows:", df.duplicated().sum())

print("\nData types:")
df.info()
```

```python
df.to_csv("cleaned_dataset.csv", index=False)

print("Cleaned dataset saved successfully!")
```

## 2. Clean duplicate rows
This removes exact duplicate rows and keeps the first occurrence.

```python
rows_before = len(df)
df = df.drop_duplicates().copy()
print('Duplicate rows removed:', rows_before - len(df))
```

## 3. Review and handle missing values
The example below fills numeric columns with their median and text columns with their mode. Review whether this is appropriate for each column before using it.

```python
print('Missing values before handling:')
display(df.isna().sum())

# Fill numeric columns with the median
numeric_cols = df.select_dtypes(include='number').columns
for col in numeric_cols:
    if df[col].isna().any():
        df[col] = df[col].fillna(df[col].median())

# Fill text columns with the mode when a mode exists
text_cols = df.select_dtypes(include=['object', 'string']).columns
for col in text_cols:
    if df[col].isna().any():
        modes = df[col].mode(dropna=True)
        if not modes.empty:
            df[col] = df[col].fillna(modes.iloc[0])

print('Missing values after this step:')
display(df.isna().sum())
```

## 4. Review inconsistent text values
The next cell trims leading/trailing spaces in text columns. Do not automatically lowercase or replace category values until you inspect the unique values and decide on a consistent standard.

```python
text_cols = df.select_dtypes(include=['object', 'string']).columns
for col in text_cols:
    df[col] = df[col].astype('string').str.strip()

for col in text_cols:
    print(f'\n{col}:')
    print(df[col].value_counts(dropna=False).head(20))
```
